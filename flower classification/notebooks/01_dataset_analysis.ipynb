```python
from pathlib import Path

dataset_path = Path(
    r"C:\Users\saman\OneDrive\Desktop\OG flower dataset\archive\102 flower"
)

print("Dataset path:")
print(dataset_path)

print("\nDataset exists:")
print(dataset_path.exists())
```

```python
flowers_path = dataset_path / "flowers"

train_path = flowers_path / "train"
valid_path = flowers_path / "valid"
test_path = flowers_path / "test"

print("Flowers folder exists:", flowers_path.exists())
print("Train folder exists:", train_path.exists())
print("Validation folder exists:", valid_path.exists())
print("Test folder exists:", test_path.exists())
```

```python
train_classes = train_path.iterdir()

number_of_train_classes = 0

for item in train_classes:
    if item.is_dir():
        number_of_train_classes = number_of_train_classes + 1

print("Number of training classes:", number_of_train_classes)


valid_classes = valid_path.iterdir()

number_of_valid_classes = 0

for item in valid_classes:
    if item.is_dir():
        number_of_valid_classes = number_of_valid_classes + 1

print("Number of validation classes:", number_of_valid_classes)


test_classes = test_path.iterdir()

number_of_test_classes = 0

for item in test_classes:
    if item.is_dir():
        number_of_test_classes = number_of_test_classes + 1

print("Number of test classes:", number_of_test_classes)
```

```python
train_image_count = 0

for class_folder in train_path.iterdir():
    if class_folder.is_dir():
        for image_file in class_folder.iterdir():
            if image_file.is_file():
                train_image_count = train_image_count + 1

print("Number of training images:", train_image_count)


valid_image_count = 0

for class_folder in valid_path.iterdir():
    if class_folder.is_dir():
        for image_file in class_folder.iterdir():
            if image_file.is_file():
                valid_image_count = valid_image_count + 1

print("Number of validation images:", valid_image_count)


test_image_count = 0

for class_folder in test_path.iterdir():
    if class_folder.is_dir():
        for image_file in class_folder.iterdir():
            if image_file.is_file():
                test_image_count = test_image_count + 1

print("Number of test images:", test_image_count)


total_image_count = (
    train_image_count
    + valid_image_count
    + test_image_count
)

print("Total number of images:", total_image_count)
```

```python

train_percentage = (train_image_count / total_image_count) * 100
valid_percentage = (valid_image_count / total_image_count) * 100
test_percentage = (test_image_count / total_image_count) * 100

print("Training percentage:", train_percentage)
print("Validation percentage:", valid_percentage)
print("Test percentage:", test_percentage)
```

```python

print("Images in each training class:")

for class_folder in train_path.iterdir():
    if class_folder.is_dir():
        
        image_count = 0
        
        for image_file in class_folder.iterdir():
            if image_file.is_file():
                image_count = image_count + 1
        
        print("Class", class_folder.name, ":", image_count, "images")
```

```python
class_image_counts = []

for class_folder in train_path.iterdir():
    if class_folder.is_dir():

        image_count = 0

        for image_file in class_folder.iterdir():
            if image_file.is_file():
                image_count = image_count + 1

        class_image_counts.append(image_count)


smallest_class = min(class_image_counts)
largest_class = max(class_image_counts)
average_class = sum(class_image_counts) / len(class_image_counts)

print("Number of classes:", len(class_image_counts))
print("Smallest class:", smallest_class, "images")
print("Largest class:", largest_class, "images")
print("Average images per class:", round(average_class, 2))
```

```python
import matplotlib.pyplot as plt

class_names = []
class_image_counts = []

for class_folder in train_path.iterdir():

    if class_folder.is_dir():

        image_count = 0

        for image_file in class_folder.iterdir():

            if image_file.is_file():
                image_count = image_count + 1

        class_number = int(class_folder.name)

        class_names.append(class_number)
        class_image_counts.append(image_count)

sorted_data = sorted(zip(class_names, class_image_counts))

class_names = []
class_image_counts = []

for class_number, image_count in sorted_data:
    class_names.append(class_number)
    class_image_counts.append(image_count)

# bar chart
plt.figure(figsize=(16, 6))

plt.bar(class_names, class_image_counts)

plt.xlabel("Flower Class")
plt.ylabel("Number of Training Images")
plt.title("Training Images per Flower Class")

# Show every class number
plt.xticks(class_names, rotation=90)

plt.tight_layout()

plt.show()
```

```python
image_sizes = []

for class_folder in train_path.iterdir():

    if class_folder.is_dir():

        for image_file in class_folder.iterdir():

            if image_file.is_file():

                image = Image.open(image_file)

                image_sizes.append(image.size)


smallest_width = min(size[0] for size in image_sizes)
largest_width = max(size[0] for size in image_sizes)

smallest_height = min(size[1] for size in image_sizes)
largest_height = max(size[1] for size in image_sizes)


print("Number of images checked:", len(image_sizes))
print("Smallest width:", smallest_width)
print("Largest width:", largest_width)
print("Smallest height:", smallest_height)
print("Largest height:", largest_height)
```

```python
from PIL import Image

rgb_count = 0
other_mode_count = 0

jpeg_count = 0
other_format_count = 0


for class_folder in train_path.iterdir():

    if class_folder.is_dir():

        for image_file in class_folder.iterdir():

            if image_file.is_file():

                image = Image.open(image_file)

                if image.mode == "RGB":
                    rgb_count = rgb_count + 1
                else:
                    other_mode_count = other_mode_count + 1

                if image.format == "JPEG":
                    jpeg_count = jpeg_count + 1
                else:
                    other_format_count = other_format_count + 1


print("RGB images:", rgb_count)
print("Non-RGB images:", other_mode_count)

print("JPEG images:", jpeg_count)
print("Non-JPEG images:", other_format_count)
```

```python
# number of images in each class

train_counts = []
valid_counts = []
test_counts = []

class_numbers = []


for class_folder in train_path.iterdir():

    if class_folder.is_dir():

        class_number = int(class_folder.name)

        train_count = 0

        for image_file in class_folder.iterdir():
            if image_file.is_file():
                train_count = train_count + 1


        valid_class_folder = valid_path / str(class_number)

        valid_count = 0

        for image_file in valid_class_folder.iterdir():
            if image_file.is_file():
                valid_count = valid_count + 1


        test_class_folder = test_path / str(class_number)

        test_count = 0

        for image_file in test_class_folder.iterdir():
            if image_file.is_file():
                test_count = test_count + 1


        class_numbers.append(class_number)
        train_counts.append(train_count)
        valid_counts.append(valid_count)
        test_counts.append(test_count)


sorted_data = sorted(
    zip(class_numbers, train_counts, valid_counts, test_counts)
)


class_numbers = []
train_counts = []
valid_counts = []
test_counts = []


for class_number, train_count, valid_count, test_count in sorted_data:

    class_numbers.append(class_number)
    train_counts.append(train_count)
    valid_counts.append(valid_count)
    test_counts.append(test_count)


print("Class 1:")
print("Training:", train_counts[0])
print("Validation:", valid_counts[0])
print("Test:", test_counts[0])

print()

print("Class 102:")
print("Training:", train_counts[101])
print("Validation:", valid_counts[101])
print("Test:", test_counts[101])
```
